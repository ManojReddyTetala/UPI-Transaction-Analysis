```python
# ============================================================
# STAGE 1 - CONTAINER 1
# DATASET OVERVIEW
# ============================================================

print("DATASET OVERVIEW")
print("=" * 50)
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("/content/Dataset(1).csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
rows, columns = df.shape

print(f"Number of records : {rows:,}")
print(f"Number of columns : {columns}")

print("\nDataset memory usage:")
memory_mb = df.memory_usage(deep=True).sum() / (1024 ** 2)
print(f"{memory_mb:.2f} MB")

print("\nIndex information:")
print(f"Index type        : {type(df.index).__name__}")
print(f"Index start       : {df.index.min()}")
print(f"Index end         : {df.index.max()}")
```

```python
# ============================================================
# STAGE 1 - CONTAINER 2
# COLUMN AND DATA-TYPE PROFILE
# ============================================================

print("COLUMN AND DATA-TYPE PROFILE")
print("=" * 70)

profile = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str).values,
    "Non-Null Count": df.notna().sum().values,
    "Unique Values": df.nunique(dropna=True).values
})

display(profile)
```

```python
# ============================================================
# STAGE 1 - CONTAINER 3
# RECORD INSPECTION
# ============================================================

print("FIRST 5 RECORDS")
print("=" * 50)
display(df.head())

print("\nLAST 5 RECORDS")
print("=" * 50)
display(df.tail())

print("\n5 RANDOM RECORDS")
print("=" * 50)
display(df.sample(5, random_state=42))
```

```python
# ============================================================
# STAGE 1 - CONTAINER 4
# DATA QUALITY INSPECTION
# ============================================================

print("DATA QUALITY INSPECTION")
print("=" * 60)

# Missing values
missing = df.isnull().sum()

# Duplicate rows
duplicates = df.duplicated().sum()

# Total missing cells
total_missing = missing.sum()

print(f"Total missing cells : {total_missing:,}")
print(f"Duplicate rows      : {duplicates:,}")

print("\nMissing values by column:")
missing_table = pd.DataFrame({
    "Column": missing.index,
    "Missing Values": missing.values
})

display(missing_table)

# Explicit conclusion
print("\nDATA QUALITY RESULT")
print("=" * 60)

if total_missing == 0:
    print("✓ No missing values were found.")

if duplicates == 0:
    print("✓ No duplicate rows were found.")
```

```python
# ============================================================
# STAGE 1 - CONTAINER 5
# INITIAL NUMERICAL PROFILE
# ============================================================

print("INITIAL NUMERICAL PROFILE")
print("=" * 70)

# The current dataset contains numerical fields stored as text/object
# values because several fields contain blank entries.  For Stage 1,
# we only profile these fields; the original dataframe is not modified.

numeric_data = pd.DataFrame(index=df.index)

for column in df.columns:
    converted = pd.to_numeric(df[column], errors="coerce")

    if converted.notna().any():
        numeric_data[column] = converted

numeric_columns = numeric_data.columns.tolist()

print(f"Number of numerical columns identified: {len(numeric_columns)}")

print("\nNumerical columns:")
for i, column in enumerate(numeric_columns, start=1):
    print(f"{i}. {column}")

print("\nBasic numerical summary:")

if len(numeric_columns) > 0:
    display(numeric_data[numeric_columns].describe().T)
else:
    print("No numerical columns were identified for profiling.")

print("\nNumerical profiling note:")
print("Numeric-looking fields were converted only in a temporary profiling table.")
print("The original dataframe remains unchanged for the next stage.")
```
