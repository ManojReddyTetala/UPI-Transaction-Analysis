## 1. Problem Statement

The project aims to analyze UPI transaction data to understand transaction patterns, transaction behaviour, and fraudulent transaction patterns using data analysis and visualization techniques.

The analysis focuses on transaction types, transaction amounts, merchant categories, transaction timing, devices, networks, and sender states. It also examines fraudulent and non-fraudulent transactions to identify patterns associated with fraudulent activity.

```python
# CONTAINER 0: IMPORTS AND DATASET

import pandas as pd

df = pd.read_csv("upi_transactions_2024.csv")

print("=" * 60)
print("STAGE 8: RESULTS & INTERPRETATION")
print("=" * 60)
print("Dataset loaded successfully.")
print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])
```

```python
# CONTAINER 1: CALCULATE RESULTS

total_transactions = len(df)
total_amount = df["amount (INR)"].sum()
average_amount = df["amount (INR)"].mean()
minimum_amount = df["amount (INR)"].min()
maximum_amount = df["amount (INR)"].max()

fraud_count = int(df["fraud_flag"].sum())
non_fraud_count = int((df["fraud_flag"] == 0).sum())
fraud_percentage = (fraud_count / total_transactions) * 100

transaction_type_counts = df["transaction type"].value_counts()
merchant_counts = df["merchant_category"].value_counts()
hour_counts = df["hour_of_day"].value_counts()
device_counts = df["device_type"].value_counts()
network_counts = df["network_type"].value_counts()
day_counts = df["day_of_week"].value_counts()

fraud_by_type = df.groupby("transaction type")["fraud_flag"].sum()
fraud_by_merchant = df.groupby("merchant_category")["fraud_flag"].sum()
fraud_by_hour = df.groupby("hour_of_day")["fraud_flag"].sum()
fraud_by_device = df.groupby("device_type")["fraud_flag"].sum()
fraud_by_network = df.groupby("network_type")["fraud_flag"].sum()
fraud_by_state = df.groupby("sender_state")["fraud_flag"].sum()

print("All Stage 8 results calculated successfully.")
```

```python
# CONTAINER 2: KEY RESULTS

print("=" * 60)
print("KEY RESULTS")
print("=" * 60)

print("\nOVERALL TRANSACTION RESULTS")
print("Total Transactions :", total_transactions)
print("Total Amount       : ₹", total_amount)
print("Average Amount     : ₹", round(average_amount, 2))
print("Minimum Amount     : ₹", minimum_amount)
print("Maximum Amount     : ₹", maximum_amount)

print("\nFRAUD RESULTS")
print("Fraudulent Transactions     :", fraud_count)
print("Non-Fraudulent Transactions :", non_fraud_count)
print("Fraud Percentage            :", round(fraud_percentage, 2), "%")

print("\nMOST COMMON TRANSACTION PATTERNS")
print("Transaction Type :", transaction_type_counts.idxmax(),
      "| Count:", int(transaction_type_counts.max()))
print("Merchant Category:", merchant_counts.idxmax(),
      "| Count:", int(merchant_counts.max()))
print("Most Active Hour :", int(hour_counts.idxmax()),
      "| Count:", int(hour_counts.max()))
print("Device Type      :", device_counts.idxmax(),
      "| Count:", int(device_counts.max()))
print("Network Type     :", network_counts.idxmax(),
      "| Count:", int(network_counts.max()))
print("Day of Week      :", day_counts.idxmax(),
      "| Count:", int(day_counts.max()))

print("\nFRAUD PATTERNS")
print("Transaction Type :", fraud_by_type.idxmax(),
      "| Fraud Cases:", int(fraud_by_type.max()))
print("Merchant Category:", fraud_by_merchant.idxmax(),
      "| Fraud Cases:", int(fraud_by_merchant.max()))
print("Hour             :", int(fraud_by_hour.idxmax()),
      "| Fraud Cases:", int(fraud_by_hour.max()))
print("Device Type      :", fraud_by_device.idxmax(),
      "| Fraud Cases:", int(fraud_by_device.max()))
print("Network Type     :", fraud_by_network.idxmax(),
      "| Fraud Cases:", int(fraud_by_network.max()))
print("Sender State     :", fraud_by_state.idxmax(),
      "| Fraud Cases:", int(fraud_by_state.max()))
```
