```python
# ============================================================
# STAGE 7: DATA VISUALIZATION
# ============================================================

# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("upi_transactions_2024.csv")

print("=" * 60)
print("STAGE 7: DATA VISUALIZATION")
print("=" * 60)

print("\nRequired libraries imported:")
print("1. Pandas")
print("2. Matplotlib")

print("\nDataset loaded successfully")
print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])
```

```python
transaction_type_counts = df["transaction type"].value_counts()

plt.figure(figsize=(8, 5))
transaction_type_counts.plot(kind="bar")

plt.title("Transaction Type Distribution")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
status_counts = df["transaction_status"].value_counts()

plt.figure(figsize=(7, 5))
status_counts.plot(kind="bar")

plt.title("Transaction Status Distribution")
plt.xlabel("Transaction Status")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
fraud_counts = df["fraud_flag"].map({
    0: "Non-Fraudulent",
    1: "Fraudulent"
}).value_counts()

plt.figure(figsize=(7, 5))
fraud_counts.plot(kind="bar")

plt.title("Fraudulent vs Non-Fraudulent Transactions")
plt.xlabel("Transaction Category")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
merchant_counts = df["merchant_category"].value_counts()

plt.figure(figsize=(10, 5))
merchant_counts.plot(kind="bar")

plt.title("Transactions by Merchant Category")
plt.xlabel("Merchant Category")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
hour_counts = df["hour_of_day"].value_counts().sort_index()

plt.figure(figsize=(10, 5))
hour_counts.plot(kind="line", marker="o")

plt.title("Transaction Activity by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Number of Transactions")
plt.xticks(range(24))
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
device_counts = df["device_type"].value_counts()

plt.figure(figsize=(7, 5))
device_counts.plot(kind="bar")

plt.title("Transactions by Device Type")
plt.xlabel("Device Type")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
device_counts = df["device_type"].value_counts()

plt.figure(figsize=(7, 5))
device_counts.plot(kind="bar")

plt.title("Transactions by Device Type")
plt.xlabel("Device Type")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.hist(
    df["amount (INR)"],
    bins=30
)

plt.title("Distribution of Transaction Amounts")
plt.xlabel("Transaction Amount (INR)")
plt.ylabel("Number of Transactions")
plt.tight_layout()
plt.show()
```

```python
amount_by_type = (
    df.groupby("transaction type")["amount (INR)"]
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))
amount_by_type.plot(kind="bar")

plt.title("Average Transaction Amount by Transaction Type")
plt.xlabel("Transaction Type")
plt.ylabel("Average Amount (INR)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
fraud_by_type = (
    df.groupby("transaction type")["fraud_flag"]
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))
fraud_by_type.plot(kind="bar")

plt.title("Fraudulent Transactions by Transaction Type")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Fraudulent Transactions")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
fraud_by_merchant = (
    df.groupby("merchant_category")["fraud_flag"]
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10, 5))
fraud_by_merchant.plot(kind="bar")

plt.title("Fraudulent Transactions by Merchant Category")
plt.xlabel("Merchant Category")
plt.ylabel("Number of Fraudulent Transactions")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
fraud_by_hour = (
    df.groupby("hour_of_day")["fraud_flag"]
      .sum()
      .sort_index()
)

plt.figure(figsize=(10, 5))
fraud_by_hour.plot(kind="line", marker="o")

plt.title("Fraudulent Transactions by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Number of Fraudulent Transactions")
plt.xticks(range(24))
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
fraud_by_device = (
    df.groupby("device_type")["fraud_flag"]
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(7, 5))
fraud_by_device.plot(kind="bar")

plt.title("Fraudulent Transactions by Device Type")
plt.xlabel("Device Type")
plt.ylabel("Number of Fraudulent Transactions")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
fraud_by_network = (
    df.groupby("network_type")["fraud_flag"]
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(7, 5))
fraud_by_network.plot(kind="bar")

plt.title("Fraudulent Transactions by Network Type")
plt.xlabel("Network Type")
plt.ylabel("Number of Fraudulent Transactions")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
day_counts = df["day_of_week"].value_counts()

day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

day_counts = day_counts.reindex(day_order)

plt.figure(figsize=(9, 5))
day_counts.plot(kind="bar")

plt.title("Transactions by Day of Week")
plt.xlabel("Day of Week")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
weekend_counts = df["is_weekend"].map({
    0: "Weekday",
    1: "Weekend"
}).value_counts()

plt.figure(figsize=(7, 5))
weekend_counts.plot(kind="bar")

plt.title("Weekday vs Weekend Transactions")
plt.xlabel("Day Category")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
fraud_by_state = (
    df.groupby("sender_state")["fraud_flag"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10, 5))
fraud_by_state.plot(kind="bar")

plt.title("Top 10 Sender States by Fraudulent Transactions")
plt.xlabel("Sender State")
plt.ylabel("Number of Fraudulent Transactions")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
avg_amount_fraud = (
    df.groupby("fraud_flag")["amount (INR)"]
      .mean()
)

avg_amount_fraud.index = [
    "Non-Fraudulent" if x == 0 else "Fraudulent"
    for x in avg_amount_fraud.index
]

plt.figure(figsize=(7, 5))
avg_amount_fraud.plot(kind="bar")

plt.title("Average Transaction Amount: Fraud vs Non-Fraud")
plt.xlabel("Transaction Category")
plt.ylabel("Average Amount (INR)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
