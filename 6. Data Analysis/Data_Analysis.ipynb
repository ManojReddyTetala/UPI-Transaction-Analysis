```python
import pandas as pd

df = pd.read_csv("upi_transactions_2024.csv")

print("=" * 60)
print("STAGE 6: DATA ANALYSIS")
print("=" * 60)

# 1. Overall transaction analysis
print("\n1. OVERALL TRANSACTION ANALYSIS")
print("Total transactions :", len(df))
print("Total amount       : ₹", df["amount (INR)"].sum())
print("Average amount     : ₹", round(df["amount (INR)"].mean(), 2))
print("Maximum amount     : ₹", df["amount (INR)"].max())
print("Minimum amount     : ₹", df["amount (INR)"].min())

# 2. Transaction status analysis
print("\n2. TRANSACTION STATUS ANALYSIS")

status_counts = df["transaction_status"].value_counts()

for status, count in status_counts.items():
    percentage = (count / len(df)) * 100
    print(f"{status} : {count} ({percentage:.2f}%)")

# 3. Transaction type analysis
print("\n3. TRANSACTION TYPE ANALYSIS")

type_counts = df["transaction type"].value_counts()

for transaction_type, count in type_counts.items():
    percentage = (count / len(df)) * 100
    print(f"{transaction_type} : {count} ({percentage:.2f}%)")
```

```python
# 4. FRAUD ANALYSIS

print("\n4. FRAUD ANALYSIS")

fraud_counts = df["fraud_flag"].value_counts()

total_fraud = int(df["fraud_flag"].sum())
total_non_fraud = int((df["fraud_flag"] == 0).sum())

fraud_percentage = (total_fraud / len(df)) * 100

print("Fraudulent transactions     :", total_fraud)
print("Non-fraudulent transactions :", total_non_fraud)
print("Fraud percentage            :", round(fraud_percentage, 2), "%")

# 5. Fraud by transaction type
print("\n5. FRAUD BY TRANSACTION TYPE")

fraud_by_type = (
    df.groupby("transaction type")["fraud_flag"]
      .agg(["sum", "count"])
      .reset_index()
)

for _, row in fraud_by_type.iterrows():
    rate = (row["sum"] / row["count"]) * 100
    print(
        f"{row['transaction type']} : "
        f"{int(row['sum'])} fraud cases "
        f"out of {int(row['count'])} "
        f"({rate:.2f}%)"
    )

# 6. Fraud by merchant category
print("\n6. FRAUD BY MERCHANT CATEGORY")

fraud_by_merchant = (
    df.groupby("merchant_category")["fraud_flag"]
      .agg(["sum", "count"])
      .reset_index()
      .sort_values("sum", ascending=False)
)

for _, row in fraud_by_merchant.head(5).iterrows():
    rate = (row["sum"] / row["count"]) * 100
    print(
        f"{row['merchant_category']} : "
        f"{int(row['sum'])} fraud cases "
        f"({rate:.2f}%)"
    )
```

```python
# 7. FRAUD BY HOUR

print("\n7. FRAUD BY HOUR")

fraud_by_hour = (
    df.groupby("hour_of_day")["fraud_flag"]
      .sum()
      .reset_index()
      .sort_values("fraud_flag", ascending=False)
)

for _, row in fraud_by_hour.head(5).iterrows():
    print(
        f"Hour {int(row['hour_of_day'])} : "
        f"{int(row['fraud_flag'])} fraud cases"
    )

# 8. DEVICE TYPE ANALYSIS

print("\n8. FRAUD BY DEVICE TYPE")

fraud_by_device = (
    df.groupby("device_type")["fraud_flag"]
      .agg(["sum", "count"])
      .reset_index()
)

for _, row in fraud_by_device.iterrows():
    rate = (row["sum"] / row["count"]) * 100
    print(
        f"{row['device_type']} : "
        f"{int(row['sum'])} fraud cases "
        f"({rate:.2f}%)"
    )

# 9. NETWORK TYPE ANALYSIS

print("\n9. FRAUD BY NETWORK TYPE")

fraud_by_network = (
    df.groupby("network_type")["fraud_flag"]
      .agg(["sum", "count"])
      .reset_index()
)

for _, row in fraud_by_network.iterrows():
    rate = (row["sum"] / row["count"]) * 100
    print(
        f"{row['network_type']} : "
        f"{int(row['sum'])} fraud cases "
        f"({rate:.2f}%)"
    )

# 10. FINAL ANALYSIS SUMMARY

print("\n" + "=" * 60)
print("10. ANALYSIS COMPLETED")
print("=" * 60)

print("Analysed:")
print("- Transaction amounts")
print("- Transaction status")
print("- Transaction types")
print("- Fraud rate")
print("- Fraud by transaction type")
print("- Fraud by merchant category")
print("- Fraud by hour")
print("- Fraud by device type")
print("- Fraud by network type")

print("\n" + "=" * 60)
print("STAGE 6 COMPLETED")
print("=" * 60)
```
