```python
import pandas as pd

# Load dataset
df = pd.read_csv("upi_transactions_2024.csv")

print("=" * 60)
print("STAGE 5: DATA AGGREGATION & REPRESENTATION")
print("=" * 60)

# 1. Dataset Information
print("\n1. DATASET")
print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])

# 2. Transaction Type Aggregation
transaction_type_summary = (
    df.groupby("transaction type")
      .agg(
          transaction_count=("transaction id", "count"),
          total_amount=("amount (INR)", "sum"),
          average_amount=("amount (INR)", "mean")
      )
      .reset_index()
)

print("\n2. TRANSACTION TYPE SUMMARY")

for _, row in transaction_type_summary.iterrows():
    print("\nTransaction Type :", row["transaction type"])
    print("Transaction Count:", int(row["transaction_count"]))
    print("Total Amount     :", round(row["total_amount"], 2))
    print("Average Amount   :", round(row["average_amount"], 2))

# 3. Merchant Category Aggregation
merchant_summary = (
    df.groupby("merchant_category")
      .agg(
          transaction_count=("transaction id", "count"),
          total_amount=("amount (INR)", "sum"),
          average_amount=("amount (INR)", "mean")
      )
      .reset_index()
      .sort_values("transaction_count", ascending=False)
)

print("\n3. MERCHANT CATEGORY SUMMARY")

for _, row in merchant_summary.iterrows():
    print("\nMerchant Category :", row["merchant_category"])
    print("Transaction Count :", int(row["transaction_count"]))
    print("Total Amount      :", round(row["total_amount"], 2))
    print("Average Amount    :", round(row["average_amount"], 2))
```

```python
# 4. Fraud Aggregation
fraud_summary = (
    df.groupby("fraud_flag")
      .agg(
          transaction_count=("transaction id", "count"),
          total_amount=("amount (INR)", "sum"),
          average_amount=("amount (INR)", "mean")
      )
      .reset_index()
)

print("\n4. FRAUD FLAG SUMMARY")

for _, row in fraud_summary.iterrows():
    status = "Fraudulent" if row["fraud_flag"] == 1 else "Non-Fraudulent"

    print("\nStatus           :", status)
    print("Transaction Count:", int(row["transaction_count"]))
    print("Total Amount     :", round(row["total_amount"], 2))
    print("Average Amount   :", round(row["average_amount"], 2))

# 5. State Aggregation
state_summary = (
    df.groupby("sender_state")
      .agg(
          transaction_count=("transaction id", "count"),
          total_amount=("amount (INR)", "sum")
      )
      .reset_index()
      .sort_values("transaction_count", ascending=False)
)

print("\n5. TOP 5 SENDER STATES")

for _, row in state_summary.head(5).iterrows():
    print("\nState            :", row["sender_state"])
    print("Transaction Count:", int(row["transaction_count"]))
    print("Total Amount     :", round(row["total_amount"], 2))
```

```python
# 6. Hourly Aggregation
hourly_summary = (
    df.groupby("hour_of_day")
      .agg(
          transaction_count=("transaction id", "count"),
          total_amount=("amount (INR)", "sum")
      )
      .reset_index()
      .sort_values("transaction_count", ascending=False)
)

print("\n6. TOP 5 TRANSACTION HOURS")

for _, row in hourly_summary.head(5).iterrows():
    print("\nHour             :", int(row["hour_of_day"]))
    print("Transaction Count:", int(row["transaction_count"]))
    print("Total Amount     :", round(row["total_amount"], 2))

# 7. Representation Summary
print("\n" + "=" * 60)
print("7. AGGREGATED DATASETS CREATED")
print("=" * 60)

print("Transaction Type Summary :", transaction_type_summary.shape)
print("Merchant Category Summary:", merchant_summary.shape)
print("Fraud Summary            :", fraud_summary.shape)
print("State Summary            :", state_summary.shape)
print("Hourly Summary           :", hourly_summary.shape)

# Completion
print("\n" + "=" * 60)
print("STAGE 5 COMPLETED")
print("=" * 60)
```
