# Stage 3 — Data Extraction

## Objective

The objective of this stage is to extract useful information from the transaction
dataset for subsequent validation, aggregation, analysis, and visualization.

The timestamp field is converted into a datetime format and useful temporal
attributes are extracted to make time-based analysis easier.

Relevant transaction attributes are also selected to create a focused dataset
for the subsequent stages.

## Operations Performed

1. Loaded the transaction dataset.
2. Converted the timestamp field into datetime format.
3. Extracted date, month, year, week number, day of month, and day name.
4. Selected the relevant transaction attributes.
5. Verified the extracted dataset structure.
6. Displayed the extracted fields and sample records.

## Output

A focused extraction dataset is created containing the transaction attributes
required for analysis together with useful temporal attributes derived from
the timestamp.


```python
# ============================================================
# STAGE 3 - CONTAINER 1
# SOURCE DATASET
# ============================================================

import pandas as pd
import numpy as np

df = pd.read_csv("/content/Dataset(1).csv")

print("SOURCE DATASET")
print("=" * 70)
print(f"Number of records : {df.shape[0]:,}")
print(f"Number of columns : {df.shape[1]}")

print("\nDataset columns:")
for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")
```

## Timestamp Extraction

The timestamp field contains the date and time information for each
transaction. Converting it into datetime format allows individual temporal
components to be extracted and analysed separately.

The extracted temporal attributes are:

- Date
- Month
- Year
- Week number
- Day of month
- Day name

These attributes make it possible to study transaction behaviour across
different periods and days.


```python
# ============================================================
# STAGE 3 - CONTAINER 2
# TEMPORAL ATTRIBUTE EXTRACTION
# ============================================================

df["time_stamp"] = pd.to_datetime(
    df["time_stamp"],
    errors="coerce"
)

df["date"] = df["time_stamp"].dt.date
df["month"] = df["time_stamp"].dt.month
df["year"] = df["time_stamp"].dt.year
df["week_number"] = df["time_stamp"].dt.isocalendar().week.astype("Int64")
df["day_of_month"] = df["time_stamp"].dt.day
df["day_name"] = df["time_stamp"].dt.day_name()

print("TEMPORAL ATTRIBUTES EXTRACTED")
print("=" * 70)

temporal_columns = [
    "time_stamp",
    "date",
    "month",
    "year",
    "week_number",
    "day_of_month",
    "day_name"
]

display(df[temporal_columns].head())
```

## Transaction Attribute Extraction

The transaction fields required for the analysis are selected and combined
with the temporal attributes derived from the timestamp.

This creates a focused extraction dataset that can be passed to the next
stages without repeatedly selecting the same fields.


```python
# ============================================================
# STAGE 3 - CONTAINER 3
# TRANSACTION ATTRIBUTE EXTRACTION
# ============================================================

extraction_columns = [
    " Transaction ID ",
    "time_stamp",
    "Transaction_Type",
    " merchant category ",
    "amount_INR",
    "Transaction Status",
    "sender_agegroup",
    "receiver_age_group ",
    "Sender State",
    "sender bank",
    "Receiver_Bank",
    "device type",
    "Network Type ",
    "Fraud Flag",
    "hour of day",
    "Day Of Week",
    "is weekend"
]

extracted_df = df[extraction_columns].copy()

extracted_df["date"] = df["date"]
extracted_df["month"] = df["month"]
extracted_df["year"] = df["year"]
extracted_df["week_number"] = df["week_number"]
extracted_df["day_of_month"] = df["day_of_month"]
extracted_df["day_name"] = df["day_name"]

print("EXTRACTED DATASET")
print("=" * 70)
print(f"Number of records : {extracted_df.shape[0]:,}")
print(f"Number of columns : {extracted_df.shape[1]}")
```

```python
# ============================================================
# STAGE 3 - CONTAINER 4
# EXTRACTED DATASET PROFILE
# ============================================================

extracted_profile = pd.DataFrame({
    "Column": extracted_df.columns,
    "Data Type": extracted_df.dtypes.astype(str).values,
    "Non-Null Values": extracted_df.notna().sum().values,
    "Unique Values": extracted_df.nunique(dropna=True).values
})

display(extracted_profile)
```

## Extracted Data Preview

The first few records are displayed to verify that the selected transaction
attributes and the derived temporal information are aligned with the original
transaction records.


```python
# ============================================================
# STAGE 3 - CONTAINER 5
# SAMPLE EXTRACTED TRANSACTIONS
# ============================================================

print("SAMPLE EXTRACTED TRANSACTIONS")
print("=" * 70)

display(extracted_df.head())
```

```python
# ============================================================
# STAGE 3 - CONTAINER 6
# EXTRACTION SUMMARY
# ============================================================

print("EXTRACTION SUMMARY")
print("=" * 70)

print(f"Original records : {df.shape[0]:,}")
print(f"Extracted records: {extracted_df.shape[0]:,}")
print(f"Original attributes selected: {len(extraction_columns)}")
print(f"Derived temporal attributes : 6")

print("\nDerived temporal attributes:")
for i, column in enumerate(
    ["date", "month", "year", "week_number", "day_of_month", "day_name"],
    start=1
):
    print(f"{i}. {column}")
```
