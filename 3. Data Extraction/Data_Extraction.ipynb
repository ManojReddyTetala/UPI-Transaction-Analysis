# Stage 3 — Data Extraction

## Objective

The objective of this stage is to extract the specific information required for
subsequent data validation, aggregation, analysis, and visualization.

The transaction timestamp is converted into a proper datetime format so that
useful temporal attributes such as date, month, and year can be extracted.

A focused dataframe containing the attributes required for downstream analysis
is then created from the original transaction dataset.

## Operations Performed

1. Loaded the transaction dataset.
2. Converted the timestamp column into datetime format.
3. Extracted date, month, and year from the timestamp.
4. Selected the attributes required for subsequent analysis.
5. Verified the dimensions of the extracted dataset.
6. Displayed the extracted attributes.
7. Inspected sample extracted transactions.

## Output

The result of this stage is a focused dataframe containing the transaction,
temporal, amount, status, fraud, geographical, device, network, and time-related
attributes required for the following stages.

```python
# ============================================================
# STAGE 3 — CONTAINER 1
# SOURCE DATASET
# ============================================================

import pandas as pd
import numpy as np

df = pd.read_csv("upi_transactions_2024.csv")

print("SOURCE DATASET")
print("=" * 60)

print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]}")
```

```python
# ============================================================
# STAGE 3 — CONTAINER 2
# TIMESTAMP AND TEMPORAL EXTRACTION
# ============================================================

df["timestamp"] = pd.to_datetime(
    df["timestamp"],
    errors="coerce"
)

df["date"] = df["timestamp"].dt.date
df["month"] = df["timestamp"].dt.month
df["year"] = df["timestamp"].dt.year

print("Temporal extraction completed.")
print("=" * 60)

print("Original temporal field : timestamp")
print("Extracted fields        : date, month, year")

print("\nSample temporal information:")
display(
    df[
        ["timestamp", "date", "month", "year"]
    ].head()
)
```

```python
# ============================================================
# STAGE 3 — CONTAINER 3
# REQUIRED ATTRIBUTE EXTRACTION
# ============================================================

required_columns = [
    "transaction id",
    "timestamp",
    "date",
    "month",
    "year",
    "transaction type",
    "merchant_category",
    "amount (INR)",
    "transaction_status",
    "fraud_flag",
    "sender_state",
    "device_type",
    "network_type",
    "hour_of_day",
    "day_of_week",
    "is_weekend"
]

extracted_df = df[required_columns].copy()

print("REQUIRED ATTRIBUTES EXTRACTED")
print("=" * 60)

print(f"Extracted rows    : {extracted_df.shape[0]:,}")
print(f"Extracted columns : {extracted_df.shape[1]}")
```

```python
# ============================================================
# STAGE 3 — CONTAINER 4
# EXTRACTED DATASET PROFILE
# ============================================================

extracted_profile = pd.DataFrame({
    "Column": extracted_df.columns,
    "Data Type": extracted_df.dtypes.astype(str).values,
    "Non-Null Values": extracted_df.notna().sum().values,
    "Unique Values": extracted_df.nunique(dropna=True).values
})

display(extracted_profile)
```

```python
# ============================================================
# STAGE 3 — CONTAINER 4
# EXTRACTED DATASET PROFILE
# ============================================================

extracted_profile = pd.DataFrame({
    "Column": extracted_df.columns,
    "Data Type": extracted_df.dtypes.astype(str).values,
    "Non-Null Values": extracted_df.notna().sum().values,
    "Unique Values": extracted_df.nunique(dropna=True).values
})

display(extracted_profile)
```

```python
# ============================================================
# STAGE 3 — CONTAINER 5
# SAMPLE EXTRACTED TRANSACTIONS
# ============================================================

print("SAMPLE EXTRACTED TRANSACTIONS")
print("=" * 60)

display(
    extracted_df.head(3)
)
```
